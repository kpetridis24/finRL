```python
import datetime as dt, gzip, json, os, time, requests
from pathlib import Path
from dateutil import tz

API_KEY = "2de8304f-9a22-4787-a5d3-5b3e070640e7"
BASE_URL = "https://rest.coinapi.io/v1/orderbooks"

def _iso_z(ts: dt.datetime) -> str:
    return ts.strftime("%Y-%m-%dT%H:%M:%SZ")

def _fetch_hour(symbol_id: str,
                t_start: dt.datetime,
                levels: int,
                retries: int = 5,
                connect_to: int = 15,
                read_to: int = 120):
    params = {
        "time_start": _iso_z(t_start),
        "time_end":   _iso_z(t_start + dt.timedelta(hours=1)),
        "limit_levels": levels
    }
    headers = {"X-CoinAPI-Key": API_KEY}
    url = f"{BASE_URL}/{symbol_id}/history"

    backoff = 1
    for _ in range(retries):
        try:
            r = requests.get(url, headers=headers,
                             params=params, timeout=(connect_to, read_to))
        except requests.exceptions.ConnectTimeout:
            time.sleep(backoff); backoff *= 2; continue

        if r.status_code == 200:
            return r.json()

        if r.status_code in {429, 500, 502, 503}:
            time.sleep(backoff); backoff = min(backoff * 2, 60); continue
        if r.status_code == 401:
            raise RuntimeError("Key not authorised for historical order-book")
        raise RuntimeError(f"HTTP {r.status_code}: {r.text[:120]}")

    raise RuntimeError("Max retries exceeded")

def run(symbol_id: str,
        levels: int,
        start: str,
        end: str,
        out_root: str = "./coinapi_depth"):
    """
    Download snapshots from `start` to `end` (inclusive, UTC date strings).

    Example call in the notebook:
        run("KRAKEN_SPOT_BTC_USD", 5, "2023-01-01", "2023-01-03")
    """
    start_date = dt.date.fromisoformat(start)
    end_date   = dt.date.fromisoformat(end)
    today_utc  = dt.datetime.utcnow().date()

    out_dir = Path(out_root) / symbol_id
    out_dir.mkdir(parents=True, exist_ok=True)

    utc = tz.UTC
    d = start_date
    while d <= end_date and d < today_utc:
        for hour in range(24):
            t0 = dt.datetime.combine(d, dt.time(hour=hour, tzinfo=utc))
            fname = out_dir / f"{d}-{hour:02d}.json.gz"
            if fname.exists():
                continue

            try:
                snaps = _fetch_hour(symbol_id, t0, levels)
            except Exception:
                print(f"Skipping: {d} {hour:02d}")
                continue

            with gzip.open(fname, "wt") as gz:
                for s in snaps:
                    gz.write(json.dumps(s) + "\n")
            print(f"{d} {hour:02d}: {len(snaps):6d} snaps saved")
            time.sleep(0.3)            # 3 req/sec keeps well below limits
        d += dt.timedelta(days=1)
    print("Done!")
```

```python
import multiprocessing as mp
from datetime import datetime, timedelta

START     = datetime(2019, 5, 12)
END       = datetime(2025, 6, 20)
OUT_ROOT  = Path("./coinapi_depth").resolve()

def split_range(start: datetime, end: datetime, parts: int):
    """Yield (sub_start, sub_end) tuples that partition [start, end] as evenly as possible."""
    total_days = (end - start).days + 1
    base = total_days // parts
    extra = total_days % parts          # first *extra* slices get one extra day

    cursor = start
    for i in range(parts):
        days = base + (1 if i < extra else 0)
        sub_start = cursor
        sub_end   = cursor + timedelta(days=days - 1)
        yield sub_start, sub_end
        cursor = sub_end + timedelta(days=1)

def worker(args):
    sub_start, sub_end = args
    run(
        symbol_id="KRAKEN_SPOT_BTC_USD",
        levels=5,
        start=sub_start.strftime("%Y-%m-%d"),
        end=sub_end.strftime("%Y-%m-%d"),
        out_root=str(OUT_ROOT)
    )

n_cpus = mp.cpu_count()
print(f"Using {n_cpus} parallel workers")
# OUT_ROOT.mkdir(parents=True, exist_ok=True)

slices = list(split_range(START, END, n_cpus))
with mp.Pool(processes=n_cpus) as pool:
    pool.map(worker, slices)

# run(
#     symbol_id="KRAKEN_SPOT_BTC_USD",
#     levels=5,
#     start="2019-04-07",
#     end="2025-06-20",        # one day demo
#     out_root="./coinapi_depth"
# )
```

```python
import gzip, json, glob, pandas as pd
from pathlib import Path

DATA_DIR   = Path("./coinapi_depth/KRAKEN_SPOT_BTC_USD")   # <-- change if needed
TOP_LEVELS = 5                                             # how many book levels to flatten

def flatten_snapshot(rec: dict, L: int = TOP_LEVELS):
    """
    Convert a CoinAPI snapshot into 40 flat columns:
    bidPrice1…5, askPrice1…5, bidVol1…5, askVol1…5
    """
    bids = rec["bids"][:L]
    asks = rec["asks"][:L]
    flat = (
        [b["price"] for b in bids] +
        [a["price"] for a in asks] +
        [b["size"]  for b in bids] +
        [a["size"]  for a in asks]
    )
    return flat

# --- 1. stream all gzip files into a list of dicts -------------------------
rows_raw   = []          # raw records
rows_flat  = []          # flattened 40-column rows
for fname in sorted(glob.glob(str(DATA_DIR / "*.json.gz"))):
    with gzip.open(fname, "rt") as gz:
        for line in gz:
            rec = json.loads(line)
            rows_raw.append(rec)
            rows_flat.append(
                [rec["time_exchange"]] + flatten_snapshot(rec)
            )

print("Loaded", len(rows_raw), "snapshots")

# --- 2A. DataFrame with nested columns -------------------------------------
df_raw = pd.DataFrame(rows_raw)
df_raw.head()
```
