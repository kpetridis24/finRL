```python
import datetime as dt
import io
import zipfile
import requests
import pandas as pd

def fetch_binance_depth(symbol: str, date: dt.date, levels: int = 10):
    """
    Download one day of Binance spot order-book snapshots (top N levels).

    Returns
    -------
    pd.DataFrame
        Columns: bids/asks price/qty for each level + lastUpdateId.
    """
    # 1) Build URL
    base = ("https://data.binance.vision/data/spot/daily/bookDepth/"
            f"{levels}/{symbol}/{symbol}-bookDepth-{levels}-{date}.zip")

    # 2) Download
    r = requests.get(base, timeout=60)
    if r.status_code != 200:
        raise ValueError(f"HTTP {r.status_code} – file probably missing: {base}")

    if "zip" not in r.headers.get("Content-Type", ""):
        raise ValueError("Expected a ZIP file; got "
                         f"{r.headers.get('Content-Type')} (bad path or date?)")

    # 3) Read ZIP into memory
    with zipfile.ZipFile(io.BytesIO(r.content)) as zf:
        csv_name = zf.namelist()[0]        # there is exactly one file inside
        with zf.open(csv_name) as f:
            df = pd.read_csv(f)

    return df

# ---------------------------------------------------------------------
# Example usage
sym  = "BTCUSDT"
date = dt.date(2025, 7, 10)               # adjust to a date that exists
depth = 10

df_raw = fetch_binance_depth(sym, date, depth)
print("Columns:", df_raw.columns.tolist()[:15], "…")
print("Rows   :", len(df_raw))

# ---------------------------------------------------------------------
# Flatten first snapshot into 40 paper-style LOB columns
def flatten_snapshot(row, levels=5):
    bids = [(float(p), float(q)) for p, q in
            (s.split(":") for s in row["bids"].split("|"))][:levels]
    asks = [(float(p), float(q)) for p, q in
            (s.split(":") for s in row["asks"].split("|"))][:levels]

    prices_sizes = (
        [p for p, _ in bids] +
        [p for p, _ in asks] +
        [q for _, q in bids] +
        [q for _, q in asks]
    )
    columns = ([f"bidPrice{i}"  for i in range(1, levels+1)] +
               [f"askPrice{i}"  for i in range(1, levels+1)] +
               [f"bidVolume{i}" for i in range(1, levels+1)] +
               [f"askVolume{i}" for i in range(1, levels+1)])
    return pd.Series(prices_sizes, index=columns)

snap_flat = flatten_snapshot(df_raw.iloc[0], levels=5)
print(snap_flat.head(10))
```

```python
import re, requests, datetime as dt
from bs4 import BeautifulSoup   # pip install beautifulsoup4

def latest_available_date(symbol: str, levels: int = 10) -> dt.date:
    index_url = (f"https://data.binance.vision/data/spot/daily/bookDepth/"
                 f"{levels}/{symbol}/")
    soup = BeautifulSoup(requests.get(index_url, timeout=30).text, "html.parser")
    zips = [a.text for a in soup.find_all("a") if a.text.endswith(".zip")]
    if not zips:
        raise RuntimeError(f"No {levels}-level depth files for {symbol}")
    # extract YYYY-MM-DD, sort, return max
    dates = sorted(re.search(r"(\d{4}-\d{2}-\d{2})", z).group(1) for z in zips)
    return dt.date.fromisoformat(dates[-1])

sym = "BTCUSDT"
depth = 6
d_available = latest_available_date(sym, depth)
print("Latest file on Vision:", d_available)   # e.g. 2024-12-28
```

```python
"""
Download historical L2 order-book data from CoinAPI and save
one compressed JSONL file per hour.

Usage  (pull 2024-12-28, Kraken spot BTC/USD, top-10 levels):
    python coinapi_l2_fetch.py 2024-12-28 KRAKEN_SPOT_BTC_USD 10
"""
import sys, datetime as dt, requests, time, json, gzip, os
from dateutil import tz, parser as dparse

def fmt(ts: dt.datetime) -> str:         # 👈 helper
    return ts.replace(tzinfo=None).isoformat(timespec="seconds") + "Z"
    # or simply: return ts.strftime("%Y-%m-%dT%H:%M:%SZ")


BASE = "https://rest.coinapi.io/v1/orderbooks"

def fetch_hour(symbol_id: str,
               t_start: dt.datetime,
               levels: int | None = None,
               retries: int = 5):

    def fmt(ts):
        return ts.replace(tzinfo=None).isoformat(timespec="seconds") + "Z"

    params = {"time_start": fmt(t_start),
              "time_end"  : fmt(t_start + dt.timedelta(hours=1))}
    if levels:
        params["limit_levels"] = levels

    url = f"{BASE}/{symbol_id}/history"
    headers = {"X-CoinAPI-Key": API_KEY}
    backoff = 1
    for _ in range(retries):
        r = requests.get(url, headers=headers, params=params, timeout=(5, 60))
        if r.status_code == 200:
            return r.json()
        if r.status_code in {429, 500, 502, 503}:
            time.sleep(backoff); backoff *= 2; continue
        raise RuntimeError(r.status_code, r.text[:200])
    raise RuntimeError("max retries exceeded")

def main(date_str: str, symbol_id: str, levels: int = 10):
    date = dt.date.fromisoformat(date_str)
    out_dir = f"./coinapi_depth/{symbol_id}"
    os.makedirs(out_dir, exist_ok=True)

    utc = tz.UTC
    for h in range(24):
        t0 = dt.datetime.combine(date, dt.time(hour=h, tzinfo=utc))
        fname = f"{out_dir}/{date}-{h:02d}.json.gz"
        if os.path.exists(fname):
            continue
        try:
            snaps = fetch_hour(symbol_id, t0, levels)
        except Exception:
            continue

        with gzip.open(fname, "wt") as gz:
            for s in snaps:
                gz.write(json.dumps(s) + "\n")
        print(f"saved {fname} ({len(snaps)} snapshots)")
        time.sleep(0.25)      # be polite: 4 req/s << free-tier burst cap


main("2025-05-10", "KRAKEN_SPOT_BTC_USD", 10)
```

```python
import datetime as dt, gzip, json, os, time, requests
from pathlib import Path
from dateutil import tz

BASE_URL = "https://rest.coinapi.io/v1/orderbooks"

def _iso_z(ts: dt.datetime) -> str:
    return ts.strftime("%Y-%m-%dT%H:%M:%SZ")

def _fetch_hour(symbol_id: str,
                t_start: dt.datetime,
                levels: int,
                retries: int = 5,
                connect_to: int = 15,
                read_to: int = 120):
    params = {
        "time_start": _iso_z(t_start),
        "time_end":   _iso_z(t_start + dt.timedelta(hours=1)),
        "limit_levels": levels
    }
    headers = {"X-CoinAPI-Key": API_KEY}
    url = f"{BASE_URL}/{symbol_id}/history"

    backoff = 1
    for _ in range(retries):
        try:
            r = requests.get(url, headers=headers,
                             params=params, timeout=(connect_to, read_to))
        except requests.exceptions.ConnectTimeout:
            time.sleep(backoff); backoff *= 2; continue

        if r.status_code == 200:
            return r.json()

        if r.status_code in {429, 500, 502, 503}:
            time.sleep(backoff); backoff = min(backoff * 2, 60); continue
        if r.status_code == 401:
            raise RuntimeError("Key not authorised for historical order-book")
        raise RuntimeError(f"HTTP {r.status_code}: {r.text[:120]}")

    raise RuntimeError("Max retries exceeded")

def run(symbol_id: str,
        levels: int,
        start: str,
        end: str,
        out_root: str = "./coinapi_depth"):
    """
    Download snapshots from `start` to `end` (inclusive, UTC date strings).

    Example call in the notebook:
        run("KRAKEN_SPOT_BTC_USD", 5, "2023-01-01", "2023-01-03")
    """
    start_date = dt.date.fromisoformat(start)
    end_date   = dt.date.fromisoformat(end)
    today_utc  = dt.datetime.utcnow().date()

    out_dir = Path(out_root) / symbol_id
    out_dir.mkdir(parents=True, exist_ok=True)

    utc = tz.UTC
    d = start_date
    while d <= end_date and d < today_utc:
        for hour in range(24):
            t0 = dt.datetime.combine(d, dt.time(hour=hour, tzinfo=utc))
            fname = out_dir / f"{d}-{hour:02d}.json.gz"
            if fname.exists():
                continue

            try:
                snaps = _fetch_hour(symbol_id, t0, levels)
            except Exception:
                continue

            with gzip.open(fname, "wt") as gz:
                for s in snaps:
                    gz.write(json.dumps(s) + "\n")
            print(f"{d} {hour:02d}: {len(snaps):6d} snaps saved")
            time.sleep(0.3)            # 3 req/sec keeps well below limits
        d += dt.timedelta(days=1)
    print("Done!")
```

```python
import gzip, json, glob, pandas as pd
from pathlib import Path

DATA_DIR   = Path("./coinapi_depth/KRAKEN_SPOT_BTC_USD")   # <-- change if needed
TOP_LEVELS = 5                                             # how many book levels to flatten

def flatten_snapshot(rec: dict, L: int = TOP_LEVELS):
    """
    Convert a CoinAPI snapshot into 40 flat columns:
    bidPrice1…5, askPrice1…5, bidVol1…5, askVol1…5
    """
    bids = rec["bids"][:L]
    asks = rec["asks"][:L]
    flat = (
        [b["price"] for b in bids] +
        [a["price"] for a in asks] +
        [b["size"]  for b in bids] +
        [a["size"]  for a in asks]
    )
    return flat

# --- 1. stream all gzip files into a list of dicts -------------------------
rows_raw   = []          # raw records
rows_flat  = []          # flattened 40-column rows
for fname in sorted(glob.glob(str(DATA_DIR / "*.json.gz"))):
    with gzip.open(fname, "rt") as gz:
        for line in gz:
            rec = json.loads(line)
            rows_raw.append(rec)
            rows_flat.append(
                [rec["time_exchange"]] + flatten_snapshot(rec)
            )

print("Loaded", len(rows_raw), "snapshots")

# --- 2A. DataFrame with nested columns -------------------------------------
df_raw = pd.DataFrame(rows_raw)
df_raw.head()
```
